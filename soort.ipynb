```python
def haversine_np(lat1, lon1, lat2, lon2):
    R = 6371.0  # Earth radius in KM
    # Convert each input independently to handle both Pandas Series and scalar inputs
    lat1, lon1, lat2, lon2 = map(np.radians, [lat1, lon1, lat2, lon2])
    
    dlat = lat2 - lat1
    dlon = lon2 - lon1
    a = np.sin(dlat / 2.0) ** 2 + np.cos(lat1) * np.cos(lat2) * np.sin(dlon / 2.0) ** 2
    return R * 2 * np.arctan2(np.sqrt(a), np.sqrt(1.0 - a))

def bearing_np(lat1, lon1, lat2, lon2):
    # Convert each input independently to handle both Pandas Series and scalar inputs
    lat1, lon1, lat2, lon2 = map(np.radians, [lat1, lon1, lat2, lon2])
    
    dlon = lon2 - lon1
    x = np.sin(dlon) * np.cos(lat2)
    y = np.cos(lat1) * np.sin(lat2) - np.sin(lat1) * np.cos(lat2) * np.cos(dlon)
    bearing = np.degrees(np.arctan2(x, y))
    return (bearing + 360) % 360

# ---------------------------------------------------------
# 1. Load Raw CSV & Parse Datetime
# ---------------------------------------------------------
ds["pickup_datetime"] = pd.to_datetime(ds["pickup_datetime"])

# ---------------------------------------------------------
# 2. Filtering & Cleaning
# ---------------------------------------------------------
ds = ds[
    (ds["passenger_count"].between(1, 6)) &
    (ds["trip_duration"].between(60, 10800)) &
    (ds["pickup_latitude"].between(40.4, 41.0)) &
    (ds["pickup_longitude"].between(-74.3, -73.6)) &
    (ds["dropoff_latitude"].between(40.4, 41.0)) &
    (ds["dropoff_longitude"].between(-74.3, -73.6))
].copy()

# ---------------------------------------------------------
# 3. Feature Engineering
# ---------------------------------------------------------
# Distance & Speed Filter
ds["distance_KM"] = haversine_np(
    ds["pickup_latitude"], ds["pickup_longitude"],
    ds["dropoff_latitude"], ds["dropoff_longitude"]
)
ds["speed_km_h"] = ds["distance_KM"] / (ds["trip_duration"] / 3600.0)
ds = ds[ds["speed_km_h"] <= 100].copy()

# Time Features
ds["hours"] = ds["pickup_datetime"].dt.hour
ds["month"] = ds["pickup_datetime"].dt.month
# Matching Spark's dayofweek: 1 (Sun) to 7 (Sat)
ds["day_of_week"] = (ds["pickup_datetime"].dt.dayofweek + 1) % 7 + 1 
ds["is_weekend"] = np.where(ds["day_of_week"] >= 6, 1, 0)
ds["is_rush_hour"] = np.where(ds["hours"].between(8, 17), 1, 0)
ds["hour_day_interaction"] = ds["hours"] * ds["day_of_week"]

# Spatial Features
ds["bearing"] = bearing_np(
    ds["pickup_latitude"], ds["pickup_longitude"],
    ds["dropoff_latitude"], ds["dropoff_longitude"]
)
ds["manhattan_distance"] = (
    (ds["pickup_latitude"] - ds["dropoff_latitude"]).abs() + 
    (ds["pickup_longitude"] - ds["dropoff_longitude"]).abs()
)

# Airport & Center Distances
jfk_lat, jfk_lon = 40.6413, -73.7781
lga_lat, lga_lon = 40.7769, -73.8740
manhattan_lat, manhattan_long = 40.7580, -73.9855

ds["pickup_dist_jfk"] = haversine_np(ds["pickup_latitude"], ds["pickup_longitude"], jfk_lat, jfk_lon)
ds["dropoff_dist_jfk"] = haversine_np(ds["dropoff_latitude"], ds["dropoff_longitude"], jfk_lat, jfk_lon)
ds["pickup_dist_lga"] = haversine_np(ds["pickup_latitude"], ds["pickup_longitude"], lga_lat, lga_lon)
ds["dropoff_dist_lga"] = haversine_np(ds["dropoff_latitude"], ds["dropoff_longitude"], lga_lat, lga_lon)

ds["is_jfk"] = np.where((ds["pickup_dist_jfk"] <= 2) | (ds["dropoff_dist_jfk"] <= 2), 1, 0)
ds["is_lga"] = np.where((ds["pickup_dist_lga"] <= 2) | (ds["dropoff_dist_lga"] <= 2), 1, 0)
ds["is_airport"] = np.where((ds["is_jfk"] == 1) | (ds["is_lga"] == 1), 1, 0)

ds["pickup_distance_from_center"] = haversine_np(ds["pickup_latitude"], ds["pickup_longitude"], manhattan_lat, manhattan_long)
ds["dropoff_distance_from_center"] = haversine_np(ds["dropoff_latitude"], ds["dropoff_longitude"], manhattan_lat, manhattan_long)

ds['Predicted']=0
# Target Variable Transformation
ds["log_duration"] = np.log1p(ds["trip_duration"])
```
